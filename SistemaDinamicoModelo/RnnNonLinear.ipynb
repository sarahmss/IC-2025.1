```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Ler os dados exportados
df = pd.read_csv('NonLinearData.csv')

# 2. Separar tempo, preditores e alvos
t = df['tempo_s']
PREDICTORS = ['phi_d', 'phi_e']
TARGETS = ['x_m', 'y_m', 'theta_rad']

# 3. Criar figura com subplots
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(15, 8), sharex=True)
fig.suptitle('Variação não linear de A(t) e f(t)', fontsize=16)

# 4. Plotar preditores
for col in PREDICTORS:
    ax1.plot(t, df[col], label=col)
ax1.set_ylabel(r'Entradas: $\phi_d$, $\phi_e$')
ax1.set_title('Entradas vs. Tempo')
ax1.grid(True)
ax1.legend()

# 5. Plotar alvos (TARGETS)
for col in TARGETS:
    ax2.plot(t, df[col], label=col)
ax2.set_xlabel('Tempo [s]')
ax2.set_ylabel(r'Saidas: $x$, $y$, $\theta$')
ax2.set_title(r'Saidas vs. Tempo')
ax2.grid(True)
ax2.legend()

# 6. Mostrar gráfico
plt.tight_layout()
plt.show()
```
