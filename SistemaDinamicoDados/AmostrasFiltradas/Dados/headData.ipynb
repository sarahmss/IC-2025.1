```python
import pandas as pd
import numpy as np

def load_data(file_path):
    colunas = ["x(Wd,We)","y(Wd,We)","theta(Wd,We)","Wd_true","We_true","Wd","We","tempo"]
    df = pd.read_csv(file_path, header=None, names=colunas)
    return df

Data = load_data("./DataSmooth4.csv")
print(Data.head())
Data.to_csv("DataSmoothSpline4.csv")
print("Data saved to DataSmoothSpline.csv")
```
